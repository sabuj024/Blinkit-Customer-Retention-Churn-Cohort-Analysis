```python
# Clean the column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")

df.columns
```

```python
# Remove duplicates if esist

df= df.drop_duplicates()
```

```python
# Check for duplicates

df.duplicated().sum()
```

```python
# Clean text and category columns

text_columns = [
     "city",
     "state",
     "customer_segment",
     "acquisition_channel",
     "device_type",
     "order_status",
     "payment_method",
     "payment_status",
     "category",
]

for col in text_columns: df[col] = df[col].astype("string").str.strip()
```

```python
# category value consistant

df["city"] = df["city"].str.title()
df["category"] = df["category"].str.title()
df["state"] = df["state"].str.title()

df["payment_method"] = df["payment_method"].replace({
    "UPI": "UPI",
    "upi": "UPI",
    "Credit Card": "Credit Card",
    "credit card": "Credit Card",
    "Debit Card": "Debit Card",
    "debit card": "Debit Card",
    "Net Banking": "Net Banking",
    "net banking": "Net Banking",
    "Cash on Delivery": "COD",
    "Cod": "COD"
})
```

```python
# Date columns into proper datetime format

date_columns = [
    "signup_date",
    "order_date",
    "order_time",
    "last_login_date",
    "churn_date"
]

for col in date_columns: df[col] = pd.to_datetime(df[col],dayfirst=True, errors="coerce")
```

```python
df[date_columns].dtypes
```

```python
# check missing values

df.isnull().sum()
```

```python
category_columns = [
    "customer_segment",
    "acquisition_channel",
    "payment_method",
    "cancel_reason"
]

for col in category_columns:
    df[col] = df[col].fillna("Unknown")
```

```python
# fix numarical values

# fix numerical values
df.loc[df["quantity"] <= 0, "quantity"] = np.nan
df.loc[df["delivery_time_minutes"] < 0, "delivery_time_minutes"] = np.nan
```

```python
df.loc[(df["order_rating"] < 1) | (df["order_rating"] >5), "order_rating"] =np.nan
```
